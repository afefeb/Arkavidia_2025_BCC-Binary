```python
import math

def entropy(*fractions):
    return -sum(f * math.log2(f) for f in fractions if f > 0)

# Contoh penggunaan
fractions = [2/7, 5/7]  # Masukkan pecahan sesuai kebutuhan
result = entropy(*fractions)
print(f"Entropi: {result:.3f}")
```
